```python
import numpy as np
import pandas as pd
import joblib
#import matplotlib.pyplot as plt
#import seaborn as sns
```

```python
df = pd.read_csv('Crop_recommendation.xls')
df = pd.DataFrame(df)
df.sample(5)
```

```python
df.shape
```

```python
df['label'].unique()
```

```python
labels = df['label'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['label'] = le.fit_transform(df['label'])
```

```python
df.sample(5)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.iloc[:,:-1], df.iloc[:,-1], test_size=0.2, random_state=42)
#x_test
```

```python
model = RandomForestClassifier(n_estimators=75, random_state=42)
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
model.feature_importances_
```

```python
user_input = input('Enter values (N,P,K,temp,hum,ph,rain): ')
inp_data = [[float(i) for i in user_input.split(',')]]
print(f'Recommended crop: {le.inverse_transform(model.predict(inp_data))[0]}')
```
